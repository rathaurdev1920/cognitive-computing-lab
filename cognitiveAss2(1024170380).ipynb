## Q1. LIST OPERATIONS

```python
import random
from collections import Counter

roll_number = "1024170380"

digits = [int(digit) for digit in roll_number]
L = [digit * 10 for digit in digits]

print("Original list L:")
print(L)

L.append(250)
print("After append(250):")
print(L)

L.insert(2, 350)
print("After insert(2, 350):")
print(L)

L.remove(250)
print("After remove(250):")
print(L)

removed_element = L.pop(2)
print("After pop(2):")
print(L)
print("Element removed by pop():", removed_element)

L.sort()
print("Ascending order:")
print(L)

L.sort(reverse=True)
print("Descending order:")
print(L)

print("First three elements:", L[:3])
print("Last three elements:", L[-3:])

average = sum(L) / len(L)
greater_than_average = [x for x in L if x > average]

print("Average of L:", average)
print("Elements greater than average:", greater_than_average)
```

## Q2. TUPLE OPERATIONS

```python
scores = tuple(L[:8])

print("\nScores tuple:")
print(scores)

highest_score = max(scores)
highest_index = scores.index(highest_score)

lowest_score = min(scores)
lowest_count = scores.count(lowest_score)

print("\nHighest score:", highest_score)
print("Index of highest score:", highest_index)

print("Lowest score:", lowest_score)
print("Number of times lowest score appears:", lowest_count)

reversed_scores = list(reversed(scores))
print("\nReversed tuple as a list:")
print(reversed_scores)

search_score = 12

if search_score in scores:
    print("First occurrence index:", scores.index(search_score))
else:
    print("Score is not present in the tuple.")

print("\nAttempting to change scores[0]...")

try:
    scores[0] = 100
except TypeError as e:
    print("Error raised:", e)

first_score, second_score, *remaining_scores = scores

print("\nFirst score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

## Q3. RANDOM NUMBERS

```python
random.seed(int(roll_number))

random_numbers = [random.randint(100, 900) for _ in range(100)]

print("\nRandom list of 100 numbers:")
print(random_numbers)

odd_numbers = [x for x in random_numbers if x % 2 != 0]

print("\nNumber of odd numbers:", len(odd_numbers))
print("Odd numbers:")
print(odd_numbers)

even_numbers = [x for x in random_numbers if x % 2 == 0]

print("\nNumber of even numbers:", len(even_numbers))
print("Even numbers:")
print(even_numbers)

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [x for x in random_numbers if is_prime(x)]

print("\nNumber of prime numbers:", len(prime_numbers))
print("Prime numbers:")
print(prime_numbers)

frequency = Counter(random_numbers)
most_frequent_number, highest_frequency = frequency.most_common(1)[0]

print("\nMost frequently occurring number:", most_frequent_number)
print("Number of occurrences:", highest_frequency)
```

## Q4. SET OPERATIONS

```python
first_8_digits = digits[:8]

A = {digit * 7 for digit in first_8_digits}
B = {digit * 9 for digit in first_8_digits}

print("\nFirst 8 digits:")
print(first_8_digits)

print("\nSet A = digit × 7:")
print(A)

print("\nSet B = digit × 9:")
print(B)

union_set = A.union(B)
print("\nUnion of A and B:")
print(union_set)

intersection_set = A.intersection(B)
print("\nIntersection of A and B:")
print(intersection_set)

A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("\nA - B:")
print(A_minus_B)

print("\nB - A:")
print(B_minus_A)

symmetric_difference = A.symmetric_difference(B)
print("\nSymmetric difference of A and B:")
print(symmetric_difference)

print("\nIs A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

X = 7

A.discard(X)

print("Set A after discard(X):")
print(A)
```

## Q5. DICTIONARY OPERATIONS

```python
name = "Devansh Rathaur"
branch = "CSE"
age = 20
city = "Patiala"
cgpa = 7.4

my_dict = {
    "name": name,
    "roll_no": roll_number,
    "branch": branch,
    "age": age,
    "city": city
}

print("\nOriginal dictionary:")
print(my_dict)

my_dict["location"] = my_dict.pop("city")
my_dict["cgpa"] = cgpa

print("\nAfter renaming city to location and adding CGPA:")
print(my_dict)

my_dict["age"] += 1

print("\nAfter increasing age by 1:")
print(my_dict)

dict_using_pop = my_dict.copy()
dict_using_del = my_dict.copy()

removed_branch = dict_using_pop.pop("branch")

print("\nDictionary after deleting branch using pop():")
print(dict_using_pop)
print("Value returned by pop():", removed_branch)

del dict_using_del["branch"]

print("\nDictionary after deleting branch using del:")
print(dict_using_del)

print("\nKey-value pairs:")

for key, value in my_dict.items():
    print(f"{key} → {value}")

print("\nChecking for email key:")

if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key does not exist.")

friend_dict = {
    "name": "Rahul Sharma",
    "roll_no": "9999999999",
    "branch": "CSE",
    "age": 21,
    "city": "Delhi"
}

print("\nFriend dictionary:")
print(friend_dict)

merged_dict = {**my_dict, **friend_dict}

print("\nMerged dictionary:")
print(merged_dict)

string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nDictionary containing only string values:")
print(string_values)

print("\n========== ASSIGNMENT COMPLETED ==========")
```
